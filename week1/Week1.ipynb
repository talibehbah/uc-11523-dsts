```python
# Load student grades from a CSV file
grades_df = pd.read_csv("scores.csv")
grades_df.head(3)  # Display the first few rows of the DataFrame
```

```python
grades_df.info()  # Display information about the DataFrame, including data types and non-null counts
```

```python
# Summary statistics for the grades DataFrame
summary_stats = grades_df.describe()
print(summary_stats)  # Output: Summary statistics for each numeric column in the DataFrame
```

```python
# Check for missing values in the grades DataFrame
grades_df.isnull().sum()
```

```python
# Convert attendance from decimal to percentage
grades_df['attendance_percentage'] = grades_df['attendance'] * 100
print(grades_df[['attendance', 'attendance_percentage']].head())  # Output: Original and converted attendance values
```

```python
# Add a new column average_score that calculates the mean of all three subjects.
grades_df['average_score'] = grades_df[['math_score', 'physics_score', 'chemistry_score']].mean(axis=1)
print(grades_df[['average_score']].head())  # Output: The calculated average scores for each student
```

```python
#Add another column performance_rating with these categories:
# "Excellent" if average ≥ 85
# "Good" if 70 ≤ average < 85
# "Needs Improvement" if average < 70
performance_rating = []
for score in grades_df['average_score']:
    if score >= 85:
        performance_rating.append("Excellent")
    elif score >= 70:
        performance_rating.append("Good")
    else:
        performance_rating.append("Needs Improvement")

grades_df['performance_rating'] = performance_rating
print(grades_df[['average_score', 'performance_rating']].head())  # Output: Average scores and their corresponding performance ratings
```

```python
# Analysis: Find which student has the highest math score.
highest_math_score = grades_df['math_score'].max()
top_student = grades_df[grades_df['math_score'] == highest_math_score]
print(top_student[['name', 'math_score']])  # Output: Student ID and
```

```python
# Calculate the overall average or each subject.
overall_average = grades_df[['math_score', 'physics_score', 'chemistry_score']].mean()
print(overall_average)  # Output: Overall average for each subject
```

```python
# Group students by performance rating and count how many students fall into each category.
performance_counts = grades_df['performance_rating'].value_counts()
print(performance_counts)  # Output: Count of students in each performance category
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plot a bar chart of average_scoore for all students.
plt.figure(figsize=(10, 6))
sns.barplot(x='name', y='average_score', data=grades_df)
plt.title('Average Scores of Students')
plt.xlabel('Student Name')
plt.ylabel('Average Score')
plt.xticks(rotation=45)
plt.show()

```

```python
# Add a pass/fail column where students fail if any score is below 50.
grades_df['pass_fail'] = grades_df.apply(lambda row: 'Fail' if (row['math_score'] < 50 or row['physics_score'] < 50 or row['chemistry_score'] < 50) else 'Pass', axis=1)
print(grades_df[['name', 'pass_fail']].head())  # Output: Student names and their pass/fail status
```

```python
# Calculate the correlation between attendance and average_score.
correlation = grades_df['attendance_percentage'].corr(grades_df['average_score'])
print(f"Correlation between attendance and average score: {correlation:.2f}")  # Output: Correlation value between attendance and average score
```

```python
# plot a scatter plot to visualize the relationship between attendance and average_score.
# add a regression line to the scatter plot to see if there's a trend.
plt.figure(figsize=(10, 6))
sns.scatterplot(x='attendance_percentage', y='average_score', data=grades_df)
sns.regplot(x='attendance_percentage', y='average_score', data=grades_df, scatter=False)
plt.title('Attendance vs Average Score')
plt.xlabel('Attendance Percentage')
plt.ylabel('Average Score')
plt.show()
```

```python
# Median petal_width for each species in the iris dataset.
median_petal_width = df.groupby('species')['petal_width'].median()
print(median_petal_width)  # Output: Median petal width for each species    
```

```python
# Plot a scatter plot of sepal_length vs sepal_width, colored by species.
# Add shape markers for each species to differentiate them visually.
plt.figure(figsize=(10, 6))
sns.scatterplot(x='petal_length', y='petal_width', hue='species', data=df, style='species')
plt.title('Petal Length vs Petal Width by Species')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')  
plt.show()
```
